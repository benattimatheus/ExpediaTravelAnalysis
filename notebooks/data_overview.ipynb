```python
from dotenv import load_dotenv

load_dotenv()

from sqlalchemy import create_engine
import pandas as pd
import os

def connect_to_db():
    engine = create_engine(
        f"postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@"
        f"{os.getenv('DB_HOST')}:{os.getenv('DB_PORT')}/{os.getenv('DB_NAME')}"
    )
    return engine
```

```python
grouped = df.groupby("distance_group").agg({
    "is_booking": "mean",
    "distance_group": "count"
}).rename(columns={"distance_group": "volume"})

# plot
fig, ax1 = plt.subplots(figsize=(8,4))

grouped["is_booking"].plot(kind="bar", ax=ax1)
ax1.set_ylabel("Booking Rate")

ax2 = ax1.twinx()
grouped["volume"].plot(ax=ax2)
ax2.set_ylabel("Volume")

plt.title("Distance Group vs Booking Rate and Volume")
plt.xticks(rotation=0)
plt.show()
```

## Key Insights

- The dataset contains a substantial number of records, providing a solid foundation for analysis and modeling.
- The target variable (`is_booking`) is imbalanced, with a relatively low booking rate, which is expected in click vs booking datasets.
- The data pipeline successfully produced a clean and structured dataset with minimal missing values.
- A significant portion of the `orig_destination_distance` variable contains placeholder values (-1), representing unknown distances.
- The high frequency of unknown distance values suggests potential limitations in location tracking or user context, which may impact downstream analysis.

## Conclusion

Overall, the dataset is well-prepared and suitable for further exploratory analysis. While data quality is generally high, certain variables (such as distance) require careful treatment due to the presence of non-informative placeholder values. Subsequent steps will focus on understanding variable distributions and their relationship with booking behavior.
